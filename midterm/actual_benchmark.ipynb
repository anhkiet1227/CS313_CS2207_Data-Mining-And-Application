Notebook này thực hiện huấn luyện và kiểm thử thực tế trên cùng một tập dữ liệu **MNIST (60,000 ảnh Train, 10,000 ảnh Test)** 
đối với 5 mô hình đại diện cho các trường phái Machine Learning:
1. **Decision Tree (CART)** - Tree-based
2. **Linear SVM (SGDClassifier)** - Margin-based
3. **Logistic Regression (saga)** - Linear Probabilistic
4. **Random Forest (100 trees)** - Ensemble Bagging
5. **Artificial Neural Network (ANN - MLP 784-128-10)** - Neural Network



```python
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import SGDClassifier, LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
# Tải và tiền xử lý dữ liệu MNIST
(X_train_raw, y_train), (X_test_raw, y_test) = keras.datasets.mnist.load_data()
X_train = (X_train_raw.reshape(-1, 28 * 28) / 255.0).astype(np.float32)
X_test = (X_test_raw.reshape(-1, 28 * 28) / 255.0).astype(np.float32)

print(f"X_train shape: {X_train.shape}, y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}, y_test shape: {y_test.shape}")
results = []
```

```python
# 1. Huấn luyện Decision Tree (CART)
print("=== 1. Huấn luyện Decision Tree (CART) ===")
t0 = time.time()
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
t_dt = time.time() - t0

y_pred_dt = dt.predict(X_test)
acc_dt = accuracy_score(y_test, y_pred_dt)
prec_dt = precision_score(y_test, y_pred_dt, average='weighted', zero_division=0)
rec_dt = recall_score(y_test, y_pred_dt, average='weighted', zero_division=0)
f1_dt = f1_score(y_test, y_pred_dt, average='weighted')

results.append({
    "Model": "Decision Tree (CART)",
    "Test Accuracy": acc_dt,
    "Test Precision": prec_dt,
    "Test Recall": rec_dt,
    "Test F1-Score": f1_dt,
    "Training Time (s)": round(t_dt, 2),
    "Complexity / Parameters": "Single deep tree (max_depth=None)"
})
print(f"DT Hoàn tất trong {t_dt:.2f}s | Acc: {acc_dt:.4f} | Prec: {prec_dt:.4f} | Rec: {rec_dt:.4f} | F1: {f1_dt:.4f}")
```

```python
# 2. Huấn luyện Linear SVM (SGDClassifier loss='hinge')
print("=== 2. Huấn luyện Linear SVM (SGDClassifier loss='hinge') ===")
t0 = time.time()
svm = SGDClassifier(loss='hinge', max_iter=1000, tol=1e-3, random_state=42)
svm.fit(X_train, y_train)
t_svm = time.time() - t0

y_pred_svm = svm.predict(X_test)
acc_svm = accuracy_score(y_test, y_pred_svm)
prec_svm = precision_score(y_test, y_pred_svm, average='weighted', zero_division=0)
rec_svm = recall_score(y_test, y_pred_svm, average='weighted', zero_division=0)
f1_svm = f1_score(y_test, y_pred_svm, average='weighted')

results.append({
    "Model": "Linear SVM (SGDClassifier)",
    "Test Accuracy": acc_svm,
    "Test Precision": prec_svm,
    "Test Recall": rec_svm,
    "Test F1-Score": f1_svm,
    "Training Time (s)": round(t_svm, 2),
    "Complexity / Parameters": "7,850 weights"
})
print(f"SVM Hoàn tất trong {t_svm:.2f}s | Acc: {acc_svm:.4f} | Prec: {prec_svm:.4f} | Rec: {rec_svm:.4f} | F1: {f1_svm:.4f}")
```

```python
# 3. Huấn luyện Logistic Regression (saga, tol=0.01)
print("=== 3. Huấn luyện Logistic Regression (saga, tol=0.01) ===")
t0 = time.time()
lr = LogisticRegression(solver='saga', max_iter=200, tol=0.01, random_state=42)
lr.fit(X_train, y_train)
t_lr = time.time() - t0

y_pred_lr = lr.predict(X_test)
acc_lr = accuracy_score(y_test, y_pred_lr)
prec_lr = precision_score(y_test, y_pred_lr, average='weighted', zero_division=0)
rec_lr = recall_score(y_test, y_pred_lr, average='weighted', zero_division=0)
f1_lr = f1_score(y_test, y_pred_lr, average='weighted')

results.append({
    "Model": "Logistic Regression (saga)",
    "Test Accuracy": acc_lr,
    "Test Precision": prec_lr,
    "Test Recall": rec_lr,
    "Test F1-Score": f1_lr,
    "Training Time (s)": round(t_lr, 2),
    "Complexity / Parameters": "7,850 weights"
})
print(f"LR Hoàn tất trong {t_lr:.2f}s | Acc: {acc_lr:.4f} | Prec: {prec_lr:.4f} | Rec: {rec_lr:.4f} | F1: {f1_lr:.4f}")
```

```python
# 4. Huấn luyện Random Forest (100 trees)
print("=== 4. Huấn luyện Random Forest (100 trees) ===")
t0 = time.time()
rf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)
t_rf = time.time() - t0

y_pred_rf = rf.predict(X_test)
acc_rf = accuracy_score(y_test, y_pred_rf)
prec_rf = precision_score(y_test, y_pred_rf, average='weighted', zero_division=0)
rec_rf = recall_score(y_test, y_pred_rf, average='weighted', zero_division=0)
f1_rf = f1_score(y_test, y_pred_rf, average='weighted')

results.append({
    "Model": "Random Forest (100 trees)",
    "Test Accuracy": acc_rf,
    "Test Precision": prec_rf,
    "Test Recall": rec_rf,
    "Test F1-Score": f1_rf,
    "Training Time (s)": round(t_rf, 2),
    "Complexity / Parameters": "Ensemble of 100 deep trees"
})
print(f"RF Hoàn tất trong {t_rf:.2f}s | Acc: {acc_rf:.4f} | Prec: {prec_rf:.4f} | Rec: {rec_rf:.4f} | F1: {f1_rf:.4f}")
```

```python
# 5. Huấn luyện ANN (MLP 784-128-10)
print("=== 5. Huấn luyện ANN (MLP 784-128-10) ===")
tf.random.set_seed(42)
ann = keras.Sequential([
    layers.Input(shape=(784,)),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])
ann.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

t0 = time.time()
history = ann.fit(X_train, y_train, epochs=10, batch_size=128, validation_split=0.1, verbose=1)
t_ann = time.time() - t0

y_prob_ann = ann.predict(X_test)
y_pred_ann = y_prob_ann.argmax(axis=1)
acc_ann = accuracy_score(y_test, y_pred_ann)
prec_ann = precision_score(y_test, y_pred_ann, average='weighted', zero_division=0)
rec_ann = recall_score(y_test, y_pred_ann, average='weighted', zero_division=0)
f1_ann = f1_score(y_test, y_pred_ann, average='weighted')

results.append({
    "Model": "ANN (MLP 784-128-10)",
    "Test Accuracy": acc_ann,
    "Test Precision": prec_ann,
    "Test Recall": rec_ann,
    "Test F1-Score": f1_ann,
    "Training Time (s)": round(t_ann, 2),
    "Complexity / Parameters": f"{ann.count_params():,} params"
})
print(f"ANN Hoàn tất trong {t_ann:.2f}s | Acc: {acc_ann:.4f} | Prec: {prec_ann:.4f} | Rec: {rec_ann:.4f} | F1: {f1_ann:.4f}")
```

```python
# Bảng đối sánh kết quả thực nghiệm hoàn chỉnh
df_results = pd.DataFrame(results)
pd.set_option('display.float_format', lambda x: f'{x:.4f}')

print("=" * 120)
print("                                      BẢNG ĐỐI SÁNH KẾT QUẢ THỰC TẾ TRÊN TẬP TEST")
print("=" * 120)
print(df_results.to_string(index=False))

print("\n" + "=" * 53)
print("       KẾT QUẢ MÔ HÌNH XUẤT SẮC NHẤT")
print("=" * 53)
best_acc_idx = df_results['Test Accuracy'].idxmax()
best_prec_idx = df_results['Test Precision'].idxmax()
best_rec_idx = df_results['Test Recall'].idxmax()
best_f1_idx = df_results['Test F1-Score'].idxmax()

print(f"Best Model (Accuracy) : {df_results.loc[best_acc_idx, 'Model']} ({df_results.loc[best_acc_idx, 'Test Accuracy']:.4f})")
print(f"Best Model (Precision): {df_results.loc[best_prec_idx, 'Model']} ({df_results.loc[best_prec_idx, 'Test Precision']:.4f})")
print(f"Best Model (Recall)   : {df_results.loc[best_rec_idx, 'Model']} ({df_results.loc[best_rec_idx, 'Test Recall']:.4f})")
print(f"Best Model (F1-Score) : {df_results.loc[best_f1_idx, 'Model']} ({df_results.loc[best_f1_idx, 'Test F1-Score']:.4f})")
print("=" * 53)
```
